```python
import pandas as pd
import plotly.express as px

# Loads the data
df = pd.read_csv("cleaned_rental_data (1).csv")

# Filters for numeric columns
numeric_df = df.select_dtypes(include=['number'])

# Reorders columns by cost, house features, and amenities (only including necessary ones)
ordered_cols = [
    'Total Rent w/ Utilities', 'Personal Income Avg',
    'Bedrooms', 'Bathrooms', 'People in Unit',
    'Parking', 'Concierge/Doorman', 'Game room', 'roof deck', 'Gym', 'Pet-friendly',
    'Laundry facilities (no need to pay)', 'Pool', 'Mail Room', 'No Additional Amenities'
]
numeric_df = numeric_df[[col for col in ordered_cols if col in numeric_df.columns]]

# Computes the correlation matrix
correlation_matrix = numeric_df.corr().round(2)

# Resets index to be read properly
corr_df = correlation_matrix.reset_index().melt(id_vars="index")
corr_df.columns = ['Variable 1', 'Variable 2', 'Correlation']

# Plots the interactive heatmap
fig = px.imshow(
    correlation_matrix,
    text_auto=True,
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    labels=dict(color="Correlation"),
    title="Correlation Heatmap of Boston Rental Data"
)

fig.update_layout(
    width=900,
    height=800,
    xaxis_title="",
    yaxis_title="",
    xaxis_tickangle=45,
    font=dict(family="Georgia", size=11)
)

fig.show()
```

```python
# Converts the chart to html format and creates the file
fig.write_html("corr_heatmap.html")

from google.colab import files
files.download('corr_heatmap.html')
```
